```python
from typing import Literal, TypedDict

from langchain_openai import ChatOpenAI
from langgraph.graph import END, START, StateGraph
from pydantic import BaseModel, Field
```

```python
class SentimentOutputSchema(BaseModel):
    sentiment: Literal['positive', 'negative', 'neutral'] = Field(description='sentiment expressed in the text')


''' Imma use Laya decision model for diagnosis questions instead of LLMs. If youre using LLMs, you can use the below schema and then another structured model. Also instead of using Laya's SDK, I will serve the decision model on llama-server and send requests to the server instead. No particular reason to do this, I just like having all models served through llama'''

# class DiagnosisOutputSchema(BaseModel):
#     issue_type: Literal["UX", "Billing", "Performance", "Bug", "Support", "Other"] = Field(description='Category of issue mentioned')
#     urgency:  Literal["low", "medium", "high"] = Field(description='How urgent or critical the issue appears to be')
```

```python
class ReviewState(TypedDict):
    review: str
    sentiment: Literal['positive', 'negative', 'neutral']
    diagnosis: dict
    response: str
```

```python
sentiment_model = ChatOpenAI(
    model='local qwen3.5',
    base_url='http://127.0.0.1:8080/v1',
    api_key='not needed' #type: ignore
).with_structured_output(SentimentOutputSchema)

# diagnosis_model = ChatOpenAI(
#     model='local qwen3.5',
#     base_url='http://127.0.0.1:8080/v1',
#     api_key='not needed' #type: ignore
# ).with_structured_output(DiagnosisOutputSchema)

llm = ChatOpenAI(
    model='local qwen3.5',
    base_url='http://127.0.0.1:8080/v1',
    api_key='not needed' #type: ignore
)
```

```python
import requests

LAYA_URL = 'http://127.0.0.1:8081/v1/systemone'
```

```python
def find_sentiment(state: ReviewState):
    review = state['review']

    prompt = f"Return the sentiment of the following text as positive, negative or neutral.\nText: {review}"
    sentiment_pydantic = sentiment_model.invoke(prompt)

    return {'sentiment': sentiment_pydantic.sentiment}


def run_diagnosis(state: ReviewState):
    response = requests.post(
        LAYA_URL,
        json={
            'state':state['review'],
            'questions': {
                'issue_type': {
                    'type': 'choice',
                    'instructions': 'Categorise the issue type in the review',
                    'criteria': {
                        "UX": "Usability or interface problems",
                        "Billing": "Payment, charge, or refund problems",
                        "Performance": "Slowness or performance problems",
                        "Bug": "Software errors or broken functionality",
                        "Support": "Customer support problems",
                        "Other": "Issues that do not fit other categories"
                    }
                },
                'urgency': {
                    'type': 'choice',
                    'instructions': 'Determine how urgently this issue needs to be addressed',
                    'criteria': {
                        'low': 'Minor inconvenience',
                        'medium': 'Significant disruption to normal use',
                        'high': 'Critical issue requiring immediate attention'
                    }
                }
            }
        },
        timeout=30
    )
    response.raise_for_status() # Raises HTTPError if status is not 2xx
    answers = response.json()['answers']

    diagnosis = {
        'issue_type': answers['issue_type']['choice'],
        'urgency': answers['urgency']['choice']
    }
    return {'diagnosis': diagnosis}


def negative_response(state: ReviewState):
    review = state['review']
    diagnosis = state['diagnosis']

    prompt = f'''You are a customer support representative responding to a customer's review.

Write a short, natural, empathetic response that acknowledges the customer's specific complaint and expresses a sincere apology where appropriate.

Rules:
1. Use ONLY facts explicitly stated in the customer review and diagnosis. Do not invent events, actions, outcomes, or internal processes.
2. Do NOT claim that a refund was issued, an issue was reported, a team was notified, or work is underway unless the provided information explicitly confirms it.
3. Acknowledge the specific problem rather than giving generic customer-service praise.
4. Use a professional, personal, and conversational tone. Avoid exaggerated gratitude, excessive apologies, corporate jargon, and robotic phrasing.
5. Do not address the customer as 'Dear Valued Customer' or use generic greetings or sign-offs.
6. Do not invent company names, employee names, or other details. Never use placeholders.
7. Do not promise a resolution or future outcome that has not been confirmed.
8. Keep the response between 40 and 70 words.
9. Output ONLY the customer-facing response. Do not include explanations, headings, or analysis.

Customer review:
{review}

Diagnosis:
{diagnosis}

Write the response now.'''

    response = llm.invoke(prompt).content

    return {'response': response}


def positive_response(state: ReviewState):
    review = state['review']

    prompt = f'You are replying to a customer who has had a positive experience. Your goal is to respond to the customer in a kind, thankful and polite way. Try to make the response seem personal and human, and that you value them greatly as a customer. Customer review: {review}'

    response = llm.invoke(prompt).content

    return {'response': response}
```

```python
def check_condition(state: ReviewState) -> Literal['run_diagnosis', 'positive_response']:
    sentiment = state['sentiment']

    if sentiment == 'positive':
        return 'positive_response'
    else:
        return 'run_diagnosis'
```

```python
graph = StateGraph(ReviewState)

graph.add_node('find_sentiment', find_sentiment)
graph.add_node('run_diagnosis', run_diagnosis)
graph.add_node('negative_response', negative_response)
graph.add_node('positive_response', positive_response)

graph.add_edge(START, 'find_sentiment')
graph.add_conditional_edges('find_sentiment', check_condition)
graph.add_edge('run_diagnosis', 'negative_response')
graph.add_edge('negative_response', END)
graph.add_edge('positive_response', END)

workflow = graph.compile()
```
